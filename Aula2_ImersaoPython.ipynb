```python
# Dicionário de renomeação
novos_nomes = {
    'work_year': 'ano',
    'experience_level': 'senioridade',
    'employment_type': 'contrato',
    'job_title': 'cargo',
    'salary': 'salario',
    'salary_currency': 'moeda',
    'salary_in_usd': 'usd',
    'employee_residence': 'residencia',
    'remote_ratio': 'remoto',
    'company_location': 'empresa',
    'company_size': 'tamanho_empresa'
}

# Aplicando renomeação
df.rename(columns=novos_nomes, inplace=True)

# Verificando resultado
df.head()
```

```python
# O método .value_counts() serve para contar quantas vezes cada valor único aparece em uma coluna.
df['senioridade'].value_counts()
```

```python
df['contrato'].value_counts()
```

```python
df['remoto'].value_counts()
```

```python
df['tamanho_empresa'].value_counts()
```

```python
senioridade = {
    'SE': 'senior',
    'MI': 'pleno',
    'EN': 'junior',
    'EX': 'executivo'
}
df['senioridade'] = df['senioridade'].replace(senioridade)
df['senioridade'].value_counts()
```

```python
contrato = {
    'FT': 'integral',
    'PT': 'parcial',
    'CT': 'contrato',
    'FL': 'freelancer'
}
df['contrato'] = df['contrato'].replace(contrato)
df['contrato'].value_counts()
```

```python
tamanho_empresa = {
    'L': 'grande',
    'S': 'pequena',
    'M': 'media'

}
df['tamanho_empresa'] = df['tamanho_empresa'].replace(tamanho_empresa)
df['tamanho_empresa'].value_counts()
```

```python
mapa_trabalho = {
    0: 'presencial',
    100: 'remoto',
    50: 'hibrido'
}

df['remoto'] = df['remoto'].replace(mapa_trabalho)
df['remoto'].value_counts()
```

Criação de um DataFrame para validar metodos de Correção da Dase de Dados

Preenchimento com a Media

```python
import numpy as np

df_salarios = pd.DataFrame({
    'nome': ["Ana", "Bruno", "Carla", "Daniel", "Eduarda"],
    'salario': [5000, 2000, np.nan, 4000, np.nan]
})

df_salarios['salario_media'] = df_salarios['salario'].fillna(df_salarios['salario'].mean().round(2))
df_salarios
```

```python
df_salarios['salario_mediana'] = df_salarios['salario'].fillna(df_salarios['salario'].median())
df_salarios
```

Novo DataFrame de Temperatura

```python
df_temperatura = pd.DataFrame({
    'data': ['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-04', '2023-01-05'],
    'temperatura': [25, 28, np.nan, np.nan, 30]
})

df_temperatura['preenchido_ffill'] = df_temperatura['temperatura'].fillna(method='ffill')

df_temperatura
```

```python
df_temperatura = pd.DataFrame({
    'data': ['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-04', '2023-01-05'],
    'temperatura': [25, 28, np.nan, np.nan, 30]
})

df_temperatura['preenchido_bfill'] = df_temperatura['temperatura'].fillna(method='bfill')

df_temperatura
```

Novo DF, para preenchimento estatico

```python
df_cidades = pd.DataFrame({
    'nome': ['Ana', 'Paulo', 'Junior', 'Roberto', 'Laura'],
    'cidade': ['São Paulo', np.nan, 'Belo Horizonte', np.nan, 'Curitiba'],
    'temperatura': [25, 28, 27, 31, 30]
})

df_cidades['cidade_preenchida'] = df_cidades["cidade"].fillna("Nao Preenchida")

df_cidades
```
