```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd

## 1. Pré-processamento dos dados =============================================

# Calcular salário médio e contagem por país e cargo
df_agg = df_limpo.groupby(['residencia', 'cargo'])['usd'].agg(['mean', 'count']).reset_index()
df_agg.rename(columns={'mean': 'salario_medio', 'count': 'qtd_registros'}, inplace=True)

# Filtrar combinações com poucos registros (ajuste o valor conforme necessário)
df_agg = df_agg[df_agg['qtd_registros'] >= 10]

# Criar coluna de destaque
df_agg['destaque'] = df_agg['cargo'].apply(
    lambda x: 'Data Scientist' if 'Data Scientist' in x else 'Outros Cargos')

# Ordenar por salário médio
df_agg = df_agg.sort_values('salario_medio', ascending=False)

# Selecionar top países (ajuste conforme necessário)
top_paises = df_agg['residencia'].value_counts().head(15).index
df_agg = df_agg[df_agg['residencia'].isin(top_paises)]

## 2. Gráfico de Barras Interativo ============================================

fig = px.bar(
    df_agg,
    x='residencia',
    y='salario_medio',
    color='destaque',
    color_discrete_map={
        'Data Scientist': '#EF553B',  # Vermelho
        'Outros Cargos': '#636EFA'   # Azul
    },
    hover_data=['cargo', 'qtd_registros'],
    labels={
        'residencia': 'País',
        'salario_medio': 'Salário Médio (USD)',
        'destaque': 'Cargo'
    },
    title='<b>Salário Médio por País</b><br><sup>Destaque para Data Scientists</sup>',
    text='salario_medio',
    height=600
)

# Formatar valores no hover
fig.update_traces(
    texttemplate='$%{text:,.0f}',
    textposition='outside',
    hovertemplate=(
        "<b>País:</b> %{x}<br>"
        "<b>Cargo:</b> %{customdata[0]}<br>"
        "<b>Salário Médio:</b> $%{y:,.0f}<br>"
        "<b>Registros:</b> %{customdata[1]}"
    )
)

# Melhorar layout
fig.update_layout(
    uniformtext_minsize=8,
    uniformtext_mode='hide',
    xaxis_tickangle=-45,
    yaxis_tickprefix='$',
    yaxis_tickformat=',.0f',
    hoverlabel=dict(
        bgcolor="white",
        font_size=12,
        font_family="Rockwell"
    ),
    legend_title_text='Tipo de Cargo',
    plot_bgcolor='rgba(0,0,0,0)',
    paper_bgcolor='rgba(0,0,0,0)',
    margin=dict(l=20, r=20, t=80, b=20)
)

## 3. Gráfico Adicional - Salário por Cargo ===================================

# Criar figura com subplots
fig2 = make_subplots(rows=1, cols=1)

# Adicionar todos os cargos em cinza
fig2.add_trace(
    go.Bar(
        x=df_agg['cargo'],
        y=df_agg['salario_medio'],
        name='Outros Cargos',
        marker_color='lightgray',
        hovertemplate='<b>Cargo:</b> %{x}<br><b>Salário:</b> $%{y:,.0f}<extra></extra>'
    )
)

# Adicionar Data Scientists em destaque
df_ds = df_agg[df_agg['destaque'] == 'Data Scientist']
fig2.add_trace(
    go.Bar(
        x=df_ds['cargo'],
        y=df_ds['salario_medio'],
        name='Data Scientist',
        marker_color='#EF553B',
        hovertemplate='<b>Cargo:</b> %{x}<br><b>Salário:</b> $%{y:,.0f}<extra></extra>'
    )
)

# Atualizar layout
fig2.update_layout(
    title='<b>Salário Médio por Cargo</b><br><sup>Destaque para Data Scientists</sup>',
    xaxis_title='Cargo',
    yaxis_title='Salário Médio (USD)',
    xaxis_tickangle=-45,
    yaxis_tickprefix='$',
    yaxis_tickformat=',.0f',
    barmode='group',
    height=600,
    showlegend=True,
    plot_bgcolor='rgba(0,0,0,0)',
    paper_bgcolor='rgba(0,0,0,0)'
)

## 4. Mostrar os gráficos =====================================================

# Mostrar o primeiro gráfico
fig.show()

# Mostrar o segundo gráfico (se houver Data Scientists)
if not df_ds.empty:
    fig2.show()
else:
    print("Nenhum Data Scientist encontrado nos dados filtrados.")

## 5. Opção para salvar os gráficos ===========================================
# fig.write_html("salario_por_pais.html")
# fig2.write_html("salario_por_cargo.html")
```
